## Section 2: Standardize and Remap

Fix the inconsistent text in `gender` using string methods, then remap the `smoker` column (which has abbreviations that string methods alone cannot fix).

```python
# 2a. Standardize the gender column.
#     Chain .str.strip() and .str.title() to remove whitespace
#     and fix the capitalization.
#     Example: "female" becomes "Female", "FEMALE" becomes "Female".

df["gender"] = df["gender"].str.strip().str.title()


print("Unique values in 'gender' after cleaning:")
print(df["gender"].unique())
# Expected: ['Female' 'Male']
```

```python
# 2b. Remap the smoker column.
#     The column has variants like "Y", "yes", "YES", "N", "NO".
#     str.title() cannot fix "Y", it stays "Y".
#     Fill in the missing values in the dictionary below.

smoker_map = {
    "Yes": "Yes",  "yes": "Yes",  "YES": "Yes",  "Y": "Yes",
    "No":  "No",   "no":  "No",   "NO":  "No",   "N": "No",
}

df["smoker"] = df["smoker"].map(smoker_map)

print("Unique values in 'smoker' after remapping:")
print(df["smoker"].unique())
# Expected: ['No' 'Yes']  (only two values)
```

## Section 3: Handle Missing Values

Two columns have missing values: `age` and `income_usd`. For each one, decide whether to fill or drop, and explain your choice.

```python
# 3a. Fill missing values in the age column with the column median.
#     We use the median (not the mean) because it is more robust to outliers.
#     Fill in the blank.

median_age = df["age"].median()
df["age"] =  df["age"].fillna(median_age)

# After filling, convert age back to integer.
# pandas stored it as float64 because it could not hold NaN in an integer column.
df["age"] = df["age"].astype(int)

print(f"Median age used to fill: {median_age}")
print(f"age dtype after conversion: {df['age'].dtype}")
print(f"Missing age values remaining: {df['age'].isnull().sum()}")
```

```python
# 3b. Fill missing values in the income_usd column with the column median.

median_income = df["income_usd"].median()
df["income_usd"] = df["income_usd"].fillna(median_income)

print(f"Median income used to fill: {median_income}")
print(f"Missing income_usd values remaining: {df['income_usd'].isnull().sum()}")
```

```python
# 3c. Confirm: print df.isnull().sum() to verify no missing values remain.
print("Missing values after filling:")
print(df.isnull().sum())
```

## Section 4: Derive a New Column

Use `apply()` to create a new column from existing data. This is the Transform operation: you define what each value should become; pandas handles the loop.

```python
# 4a. The classify_age function below is already written.
#     Read it and understand what it returns for each age range.

def classify_age(age):
    """
    TRANSFORM: classify a respondent's age into a broad group label.

    Parameters:
        age (int) -- respondent's age in years

    Returns:
        str -- one of 'Young Adult', 'Middle Aged', or 'Older Adult'
    """
    if age < 35:
        return "Young Adult"
    elif age < 50:
        return "Middle Aged"
    else:
        return "Older Adult"


# 4b. Use .apply() to create a new column 'age_group' in df.
#     Pass classify_age to .apply(); do NOT call it with parentheses.

df["age_group"] =  df["age"].apply(classify_age)

print("First 8 rows showing age and age_group:")
print(df[["response_id", "age", "age_group"]].head(8).to_string(index=False))
```

```python
# 4c. How many respondents fall in each age group?
#     Use value_counts() to find out.

print("Respondents per age group:")
print(df["age_group"].value_counts())
```

## Section 5: Merge (Bonus, 0.25 pts)

A merge combines two DataFrames using a shared key column, the same way SQL uses JOIN. Here you will merge your cleaned dataset with a ZIP code reference table that maps each ZIP code to a neighborhood name.

The shared key column is `zip_code`. Both `df` (the cleaned health survey) and `zip_ref` (the reference table) contain it. An inner merge keeps only rows where the `zip_code` appears in both DataFrames.

```python
# 5a. Load the ZIP code reference table.
zip_ref = pd.read_csv('data2.csv')
print("ZIP reference table:")
print(zip_ref)
```

```python
# 5b. Merge df with zip_ref on the zip_code column.
#     how='inner' keeps only rows where zip_code appears in BOTH DataFrames.

shape_before = df.shape

df_merged = pd.merge(df, zip_ref, on="zip_code", how="inner")

print("Shape before merge:", shape_before)
print("Shape after merge: ", df_merged.shape)
print()
print("First 3 rows of merged result (selected columns):")
print(df_merged[["response_id", "zip_code", "neighborhood", "county"]].head(3).to_string(index=False))
```
